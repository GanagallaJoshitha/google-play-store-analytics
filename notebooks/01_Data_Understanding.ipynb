# Initial Data Quality Observations

Based on the exploration, several issues are immediately visible:

- Missing values exist in the Rating column.
- Reviews, Installs, and Price are stored as text instead of numeric values.
- Duplicate records are present.
- Some rows contain corrupted values.
- The Size column contains inconsistent units (MB, KB, "Varies with device").

These issues will be addressed in the next notebook.
